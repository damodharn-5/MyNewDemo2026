```python
# Check missing values
def check_missing(telecomdf):
    null_val = telecomdf.isnull().sum()
    percent = 100 * telecomdf.isnull().sum()/len(telecomdf)
    missing_table = pd.concat([null_val, percent], axis=1)
    col = missing_table.rename(columns = {0 : 'count', 1 : 'Rate'})
    return col


check_missing(telecomdf)
```

```python
# Check missing values
def check_missing(telecomdf):
    null_val = telecomdf.isnull().sum()
    percent = 100 * telecomdf.isnull().sum() / len(telecomdf)

    missing_table = pd.concat([null_val, percent], axis=1)

    missing_table = missing_table.rename(
        columns={0: 'count', 1: 'Rate'}
    )

    # Add % symbol to Rate
    missing_table['Rate'] = missing_table['Rate'].round(2).astype(str) + '%'

    return missing_table

check_missing(telecomdf)
```
