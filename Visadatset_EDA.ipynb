```python
Prevailing_Wage=visa_df['prevailing_wage']
# By using Where Condition

Prevailing_Wage=visa_df['prevailing_wage']
q1 =round(Prevailing_Wage.quantile(0.25),2)
q2 =round(Prevailing_Wage.quantile(0.50),2)
q3 =round(Prevailing_Wage.quantile(0.75),2)
iqr=q3-q1
lb= q1-(1.5*iqr)
ub =q3+(1.5*iqr)

wage_median=np.median(Prevailing_Wage)

con=(Prevailing_Wage<lb) | (Prevailing_Wage>ub)
true=wage_median
false=wage_median
false1=ub

l=np.where(con,true,false1)
visa_df['prevailing_wage']=l
```
